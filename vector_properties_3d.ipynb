# Шпаргалка: Свойства и операции с векторами в 3D (NumPy)

Этот ноутбук содержит основные определения, математические формулы и исполняемые примеры на Python/NumPy для работы с векторами в трехмерном пространстве.

```python
import numpy as np
```

## 1. Скалярное произведение ($\mathbf{a} \cdot \mathbf{b}$)

* **Что на выходе:** Число (скаляр).
* **Формула:** $\mathbf{a} \cdot \mathbf{b} = |\mathbf{a}| \cdot |\mathbf{b}| \cdot \cos\theta$
* **В NumPy:** `np.dot(a, b)` или `a @ b`

### Главные свойства:
* **Ортогональность (перпендикулярность):** Если $\mathbf{a} \cdot \mathbf{b} = 0$, то векторы перпендикулярны ($\theta = 90^\circ$).
* **Направление:**
  * $> 0$ — угол острый ($\theta < 90^\circ$)
  * $< 0$ — угол тупой ($\theta > 90^\circ$)

```python
a = np.array([1.0, 0.0, 0.0])
b = np.array([0.0, 2.0, 0.0])
c = np.array([1.0, 1.0, 0.0])

dot_ab = a @ b
dot_ac = a @ c

print(f"a @ b = {dot_ab} -> Ортогональны: {np.isclose(dot_ab, 0.0)}")
print(f"a @ c = {dot_ac} -> Угол острый: {dot_ac > 0}")
```

## 2. Векторное произведение ($\mathbf{a} \times \mathbf{b}$)

* **Что на выходе:** Новый **вектор**, перпендикулярный плоскости исходных векторов.
* **Модуль (длина):** $|\mathbf{a} \times \mathbf{b}| = |\mathbf{a}| \cdot |\mathbf{b}| \cdot \sin\theta$ (равен площади параллелограмма, построенного на векторах).
* **В NumPy:** `cross_product = np.cross(a, b)`

### Главные свойства:
* **Коллинеарность (параллельность):** Если длина векторного произведения равна нулю ($|\mathbf{a} \times \mathbf{b}| = 0$), то векторы коллинеарны (лежат на одной прямой или параллельны).
* **Вычисление синуса угла:**
  $$\sin\theta = \frac{|\mathbf{a} \times \mathbf{b}|}{|\mathbf{a}| \cdot |\mathbf{b}|}$$

```python
v1 = np.array([2.0, 0.0, 0.0])
v2 = np.array([0.0, 3.0, 0.0])
v3 = np.array([4.0, 0.0, 0.0])  # коллинеарен v1

# Вычисление синуса угла между v1 и v2
cross_12 = np.cross(v1, v2)
cross_len_12 = np.linalg.norm(cross_12)
sin_theta = cross_len_12 / (np.linalg.norm(v1) * np.linalg.norm(v2))

print(f"Векторное произведение v1 x v2: {cross_12}")
print(f"Площадь параллелограмма (|v1 x v2|): {cross_len_12}")
print(f"sin(theta): {sin_theta}")

# Проверка коллинеарности v1 и v3
cross_len_13 = np.linalg.norm(np.cross(v1, v3))
print(f"v1 и v3 коллинеарны: {np.isclose(cross_len_13, 0.0)}")
```

## 3. Смешанное произведение ($(\mathbf{a} \times \mathbf{b}) \cdot \mathbf{c}$)

* **Что на выходе:** Число (скаляр), равное объему параллелепипеда, построенного на трех векторах.
* **В NumPy:** `mixed = np.cross(a, b) @ c` (или через определитель матрицы: `np.linalg.det([a, b, c])`)

### Главные свойства:
* **Компланарность (принадлежность одной плоскости):** Если смешанное произведение равно нулю ($(\mathbf{a} \times \mathbf{b}) \cdot \mathbf{c} = 0$), то три вектора **компланарны** (лежат в одной плоскости).

```python
a = np.array([1.0, 2.0, 3.0])
b = np.array([2.0, 4.0, 6.0])  # коллинеарен a
c = np.array([0.0, 1.0, 1.0])

mixed_product = np.cross(a, b) @ c
is_coplanar = np.isclose(mixed_product, 0.0)

print(f"Смешанное произведение: {mixed_product}")
print(f"Векторы компланарны: {is_coplanar}")
```

## 4. Проверка на базис в 3D

Чтобы три вектора $\mathbf{a}, \mathbf{b}, \mathbf{c}$ образовали полноценный базис трехмерного пространства:
1. **Они не должны быть компланарными:** Смешанное произведение (или определитель матрицы из этих векторов) **не должно быть равно нулю** ($(\mathbf{a} \times \mathbf{b}) \cdot \mathbf{c} \neq 0$).
2. **Ортогональный базис:** Дополнительно все пары векторов должны быть перпендикулярны друг другу:
   $$\mathbf{a} \cdot \mathbf{b} = 0, \quad \mathbf{b} \cdot \mathbf{c} = 0, \quad \mathbf{a} \cdot \mathbf{c} = 0$$
3. **Ортонормированный базис:** Помимо ортогональности, каждый вектор должен иметь единичную длину ($|\mathbf{a}| = 1, |\mathbf{b}| = 1, |\mathbf{c}| = 1$).

```python
def check_basis_3d(a, b, c):
    # 1. Проверка на некомпланарность (образуют ли базис вообще)
    mixed = np.cross(a, b) @ c
    if np.isclose(mixed, 0.0):
        return "Не образуют базис (векторы компланарны)"
    
    # 2. Проверка на ортогональность
    is_orthogonal = (
        np.isclose(a @ b, 0.0) and
        np.isclose(b @ c, 0.0) and
        np.isclose(a @ c, 0.0)
    )
    
    if not is_orthogonal:
        return "Обычный (неортогональный) базис"
    
    # 3. Проверка на единичную длину (ортонормированность)
    is_unit = (
        np.isclose(np.linalg.norm(a), 1.0) and
        np.isclose(np.linalg.norm(b), 1.0) and
        np.isclose(np.linalg.norm(c), 1.0)
    )
    
    if is_unit:
        return "Ортонормированный базис"
    return "Ортогональный базис"

e1 = np.array([1.0, 0.0, 0.0])
e2 = np.array([0.0, 1.0, 0.0])
e3 = np.array([0.0, 0.0, 1.0])

print("Стандартный базис:", check_basis_3d(e1, e2, e3))
print("Масштабированный базис:", check_basis_3d(e1 * 2, e2 * 3, e3))
print("Скошенный базис:", check_basis_3d(e1, np.array([1.0, 1.0, 0.0]), e3))
```
